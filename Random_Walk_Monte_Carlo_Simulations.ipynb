# Higher Dimensions Pt.1 - Why do random walks get lost in 3D?

Imagine standing at the origin of an infinite number line. You begin flipping a fair coin that tells you where you should move. If you get tails, you take one step to the left, and if you get heads, one step to the right. This is a random walk in one dimension or more. Specifically, this is a symmetric random walk with an equal probability of moving in either direction.

We can generalize this type of walk to any number of dimensions. Randomly selecting a unit step in any axis direction. So,  in 2D, we can take a step along the x or y axis, and in 3D, we can take a step along the x, y, or z axis.

In the early 1900s, Hungarian mathematician George Polya was interested in the following question: Once we begin our random walk, are we guaranteed to eventually return to where we started? We might think that if we walk forever, we’ll surely return at some point, however long it takes. But when it comes to infinity, our intuitions can mislead us. While this is correct in the 1D and 2D cases, in 3D, it is suddenly no longer true, and in fact, it’s more likely that we never return. Our probability of returning drops from a clean 1 to 0.34

Probability of returning at least once:
- 1D: 1.0
- 2D: 1.0
- 3D: 0.3405…

“A drunk man will find his way home, but a drunk bird may get lost forever.” — Shizuo Kakutani.

Why do we have this abrupt change in behavior as we jump from 2D to 3D? Sure, as we move to higher dimensional space, we have more choices for our steps. It makes sense that it would probably take longer to return, but to never return? That’s strange.

This interesting result is known as Pólya’s Recurrence Theorem (Pólya’s Random Walk Constants).

Let’s first formalize the question a bit.

Let Sn be our location at step n. Sn is equal to the sum of each of our individual steps Xi as i ranges from x to n. Each step x is drawn uniformly among the axis-aligned unit steps. So, for a given dimension D, there are two times D possible steps, and each has an equal chance of selection.

This walk can also be represented as a Markov Chain, but you can understand it without this topic.

We are concerned with this probability: P(En exists in N : Sn = 0), where 0 is the origin. In other words, the probability of returning to the origin at least once.
*  Origin called a recurrent state ==> this probability is 1
  *  When we do return to the origin, we will begin a brand new walk. Also with a guarentee of returning. Will visit the origin an infinite number of times over an infinite time frame. This condition is both necessary and sufficient for these types of systems. So, the origin in recurrent IF AND ONLY IF the expected number of returns is infinite.
*  Origin is called a transient state ==> this probability is 0 < P < 1

***Central Limit Theorem: Proves normal distribution, not uniform distribution. Exploring this below:

Looking at some walks in 1D, we'll set a 100 walkers to each take 10 steps and record their final positions. There will be a higher concentration of walkers near the origin. Even trying with 10,000 walkers and 1,000 steps, they will be more spread out, but we will still see a bell-shaped curve (normal or gaussian distribution).

Central Limit Theorem: Sum of i.i.d. RVs w/ finite mean and varianve becomes normally distributed as the number of RVs grows.
- It tells us that the distribution of the sum of independently and identically distributed random variables with finite mean and variance ultimately converges to a normal distribution.

In 2D, getting back to the origin requires both components (on the x and y axes) to return at the exact same time step, which is difficult, but recurrent nonetheless.

Carrying this logic to 3D, is a 3D walk equivalent to 3 simultaneous 1D walks? Not exactly. An actual 3D walk has six options at each step. Two for each of the three axes.

Number of outcomes at each step:
- Actual walk: 2 * 3 = 6
  - Through the derivation we find that the return probability for the 3D walk on step 2n ends up decreasing proportionally to 1/n^(3/2). The p-series now has a power greater than 1 and converges (meaning 3D is transient).
- Naive assumption: Three simultaneous 1D walks: 2^3 = 8
  - For the purposes of recurrence, the behavior is recurrent.

So, in 3D, we expect to only return a finite number of times.

Another thing to note: We've only looked at the origin, but in 1D and 2D, every other location will be visited as well.

***Origin is recurrent --> every point is recurrent (recurrence at the origin implies recurrence everywhere)

# Higher Dimensions Pt.2 - "like 2D"

In part 1, this is all convincing algebraically. But still, why this abrupt change from 2D to 3D?

*** Higher-dimensional walks are "like" expanding 2D objects.
- Higher-dimensional walks only behave "like" 2D objects expanding over time.

What does this even mean???

1D walk: root mean squared distance (avg distance traveled) at step n = sqrt(n)
- We find that this square root relationship holds in any number of dimensions. Let Sni be the i-coordinate of the walker at step n. The expected squared distance along the i-axis will be equal to the number of steps taken along that axis. So, the expected squared total distance is n, regardless of the dimensionality of the walk.

*** Asking a related questions: How many steps do we expect to have taken when we first reach a distance r?
- E[# steps when first reach distance >= r] = ?
  - Similarly, this ends up being r^2 steps. KEY

Suppose that, optimistically, when a 3D walk reaches distance r, all of its r^2 expected steps visited distinct points with 0 repeats. But, the total number of points within this distance will grow like r^3. As time goes on, the walk will only cover an ever decreasing fraction of these points proportional to r^2 over r^3 or 1 over r.
- (optimistic # points visited when first reach distance r) / (total # points within distance r) === r^2 / r^3
- It won't be able to catch up.

---

In conclusion, the rate at which 3D walks move away from the origin is too great relative to this coverage rate. In the limit, even though a 3D walk will visit infinitely many points, it will only visit 0% of the entire state space.

*** origin is recurrent --> every point gets visited
- The origin can't be recurrent because we know as with 1D and 2D cases, recurrence would imply visiting every point and while, sure, it's possible for it to return to the origin a few times by chance, eventually it will get lost forever.

The 3D walk is "like" an expanding 2D object because they exhibit similar ratios of distance reached versus expected coverage.

# Review: McKenna's Gromov-hyperbolic Groups

A simple random walk on a number line (1D), will not result in a uniform distribution as the probability of being at the extreme ends will be lower than the middle (bell curve).

However, when a random walk is performed on a finite space with symmetric transitions (meaning the probability of moving from one state to another is the same as the probabilty of moving from the other state back), the distribution of the walk tends to converge towards a uniform distribution as the number of steps increases.

So, CONDITIONS MATTER. The specific conditions of the random walk, like the state space and transition probabilities, determine whether the distribution will converge to a uniform distribution.

1. In "Random Walks and The Uniform Measure Gromov-Hyperbolic Groups" by Benjamin McKenna:
- The study of finite Markov chains that converge to the uniform distribution is well developed.
- The study of random walks on infinite groups, and specifically the question of whether they asymptotically approach some analog of the uniform distribution, is an area of active research.
- The mixing time of random walks, particularly those that converge to the uniform distribution, is well studied in literature.
  - We can consider, for example, the simple random walk on a 3D cube, which converges to the uniform distribution quiet quickly.
  - There are periodicity problems that we will look past. Formally, the 3D cube is a bipartite graph and the random walk has subsequences that converge to the uniform distributions on each of the partite sets.
  - In this example, a random walk on the cube can be thought of as a random walk on the group, and the fact that a random walk approaches a uniform distribution can be seen as a result about properties of Ds itself.
- The rest of this paper sets out to generalize this study - that is, the study of the relationship between random walks and the uniform distribution - to infinite, finitely-generated groups. Since there is no uniform distribution on infinite sets, this paper comes up with a new benchmark against which to measure the asymptotics of a random walk.

*** Mentions "Drift"

Conclusions:
- Builds off of previous work showing that such random walks do not converge to the uniform measure on the Gromov Boundary unless the group is virtually "free".
  - The fundamental obstruction to uniformity in the limiting distribution of these random walks.
- Geometric properties of hyperbolic groups influence the distribution of random walks.
- The asymptotic behavior of these walks is tied to the group's exponential growth rate and the structure of its boundary.
- In "free" groups, random walks tend to equidistribute on the boundary. However, in more complex hyperbolic groups, this property fails, reinforcing the distinction between free and non-free cases.

*** For most Gromov-hyperbolic groups, random walks governed by symmetric, finitely-supported measures do not produce a uniform distribution on the boundary, unless the group is virtually free.

# Review: Princeton Lecture 7 on Markov Chains and Random Walks

From Princeton's Lecture Notes:
Big points:
- A Markov Chain with a finite number of states will eventually distribute probability equally across all states, regardless of its starting point, if certain conditions are met regarding the transition probabilities between states.
  - In this context, "uniform distribution" means that each state in the Markov chain has an equal probability of being reached in the long run, essentially assigning the same probability to every possible state.
    - Each "state" or a path's final position ==> point sampled.
  - For a Markov chain to converge to a uniform distribution, it needs to be "ergodic", which means that it is irreducible (can transition between any pair of states) and aperiodic (does not have a repeating pattern in state transitions).
- Informally, the mixing time of a Markov chain is the time it takes to reach "nearly uniform" distribution from any arbitrary starting distribution.
- Markov chains allow one to sample from very nontrivial sets, provided we know how to find at least one element of this set. The idea is to define a Markov chain whose state space is the same as this set. The Markov chain is such that it has a unique stationary distribution, which is uniform. We know how to find one element of the set. We do a walk according to the Markov chain with this as the starting point, and after T = O(mixing time) steps, output the node we are at. This is approximately a random sample from the set.

Transition Matrix: The matrix that defines the probabilities of moving from one state to another in a single step.

Stationary Distribution: The probability distribution that remains unchanged after applying the transition matrix, which in this case would be the uniform distribution.

Convergence Rate: How quickly the Markov chain approaches the uniform distribution.

Note that M is:
1. aperiodic because of self-loops
2. connected because every sequence can be turned into the zero vector in a finite number of transformations, i.e., every node is connected to 0.
Therefore, M is ergodic, i.e., has a unique stationary distribution. Since the uniform distribution is stationary, it follows that the stationary distribution of M is uniform.

Time Mixing:

Fact (see our remark later in our analysis of mixing time): running the M for a bit longer than the mixing time results in a distribution that is extremely close to uniform.
Thus, we get the following sampling algorithm:
1. Start with the zero vector as the initial distribution of M. 2. Run M for O(n9) time.
3. output the node at which the algorithm stops.

This results in a uniform sampling from A.
Thus Markov chains are useful for sampling from a distribution. Often, we are unable to
prove any useful bounds on the mixing time (this is the case for many Markov chains used in simulated annealing and the Metropolis algorithm of statistical physics) but nevertheless in practice the chains are found to mix rapidly. Thus they are useful even though we do not have a proof that they work.

# 1D Single Random Walk Path

```python
import numpy as np
import matplotlib.pyplot as plt

num_steps = 1000
steps = np.random.choice([-1, 1], size=num_steps)
walk = np.cumsum(steps)

plt.figure(figsize=(10, 6))
plt.plot(walk)
plt.title('1D Simple Random Walk')
plt.xlabel('Step')
plt.ylabel('Position')
plt.grid(True)
plt.show()
```

# 2D Single Random Walk Path

```python
import numpy as np
import matplotlib.pyplot as plt

num_steps = 1000
directions = np.random.choice(['up', 'down', 'left', 'right'], size=num_steps)
x, y = [0], [0]

for direction in directions:
    if direction == 'up':
        x.append(x[-1])
        y.append(y[-1] + 1)
    elif direction == 'down':
        x.append(x[-1])
        y.append(y[-1] - 1)
    elif direction == 'left':
        x.append(x[-1] - 1)
        y.append(y[-1])
    elif direction == 'right':
        x.append(x[-1] + 1)
        y.append(y[-1])

plt.figure(figsize=(10, 6))
plt.plot(x, y)
plt.scatter(x[0], y[0], color='red', label='Start', zorder=5)
plt.scatter(x[-1], y[-1], color='green', label='End', zorder=5)
plt.title('2D Simple Random Walk')
plt.xlabel('X Position')
plt.ylabel('Y Position')
plt.grid(True)
plt.legend()
plt.show()
```

# Basic 2D Random Walk Analysis

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

def random_walk_2d(steps, grid_size):
    #starting at the origion
    #by the Mean Value Theorem (MVT), we should get a normal distribution
    x, y = 0, 0
    trajectory = [(x, y)]
    for _ in range(steps):
        move = np.random.choice(['up', 'down', 'left', 'right'])

        if move == 'up' and y < grid_size:
            y += 1
        elif move == 'down' and y > -grid_size:
            y -= 1
        elif move == 'left' and x > -grid_size:
            x -= 1
        elif move == 'right' and x < grid_size:
            x += 1
        trajectory.append((x, y))
    return trajectory

def monte_carlo_simulation(n_walks, steps, grid_size):
    final_positions = []
    all_trajectories = []
    for _ in range(n_walks):
        trajectory = random_walk_2d(steps, grid_size)
        final_positions.append(trajectory[-1])
        all_trajectories.append(trajectory)
    return final_positions, all_trajectories

#bounds: number of random walks, steps per walk, bounds of the grid
n_walks = 5000
steps = 100
grid_size = 20
final_positions, all_trajectories = monte_carlo_simulation(n_walks, steps, grid_size)

x_vals, y_vals = zip(*final_positions)
plt.figure(figsize=(8, 6))
sns.kdeplot(x=x_vals, y=y_vals, cmap="Blues", fill=True)
plt.scatter(0, 0, color='red', marker='x', s=100, label="Start")
plt.title("Heatmap of Final Positions in 2D Random Walk")
plt.legend()
plt.show()

plt.figure(figsize=(6, 6))
for traj in all_trajectories[:5]:
    x_vals, y_vals = zip(*traj)
    plt.plot(x_vals, y_vals, alpha=0.6)
plt.scatter(0, 0, color='red', marker='x', s=100, label="Start")
plt.title("Sample 2D Random Walks")
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

def random_walk_triangle(steps, triangle):
    x, y = np.mean(triangle, axis=0)
    trajectory = [(x, y)]
    for _ in range(steps):
        target = triangle[np.random.choice(len(triangle))]
        x = (x + target[0]) / 2
        y = (y + target[1]) / 2
        trajectory.append((x, y))
    return trajectory

def monte_carlo_simulation(n_walks, steps, triangle):
    final_positions = []
    all_trajectories = []
    for _ in range(n_walks):
        trajectory = random_walk_triangle(steps, triangle)
        final_positions.append(trajectory[-1])
        all_trajectories.append(trajectory)
    return final_positions, all_trajectories

triangle = np.array([[0, 0], [1, 0], [0.5, np.sqrt(3)/2]])
n_walks = 5000
steps = 100
final_positions, all_trajectories = monte_carlo_simulation(n_walks, steps, triangle)

x_vals, y_vals = zip(*final_positions)
plt.figure(figsize=(8, 6))
sns.kdeplot(x=x_vals, y=y_vals, cmap="Blues", fill=True)
plt.scatter(*zip(*triangle), color='red', marker='x', s=100, label="Triangle Vertices")
plt.scatter(np.mean(triangle[:, 0]), np.mean(triangle[:, 1]), color='green', marker='o', s=100, label="Origin (Centroid)")
plt.title("Heatmap of Final Positions in 2D Triangle Random Walk")
plt.legend()
plt.show()

plt.figure(figsize=(6, 6))
for traj in all_trajectories[:5]:
    x_vals, y_vals = zip(*traj)
    plt.plot(x_vals, y_vals, alpha=0.6)
plt.scatter(*zip(*triangle), color='red', marker='x', s=100, label="Triangle Vertices")
plt.scatter(np.mean(triangle[:, 0]), np.mean(triangle[:, 1]), color='green', marker='o', s=100, label="Origin (Centroid)")
plt.title("Sample 2D Random Walks in Triangle")
plt.legend()
plt.show()
```

Estimating volume using metropolis-hasting markov chain method

```python
import numpy as np
import math

def metropolis_hastings_sample(d, num_samples, R_func):
    x = np.zeros(d)
    samples = []
    for _ in range(num_samples):
        proposal = x + np.random.uniform(-0.1, 0.1, size=d)
        if R_func(proposal):
            x = proposal
        samples.append(x.copy())
    return np.array(samples)

def unit_ball(d):
    return lambda x: np.linalg.norm(x) <= 1

def estimate_volume(d, num_samples=10000):
    radii = np.logspace(0, np.log10(1.5), num=d)
    volume_ratios = []
    for i in range(1, len(radii)):
        R_outer, R_inner = radii[i], radii[i - 1]
        samples = metropolis_hastings_sample(d, num_samples, unit_ball(d))
        inside_inner = np.sum(np.linalg.norm(samples, axis=1) <= R_inner)
        ratio = inside_inner / num_samples if inside_inner > 0 else 1e-10
        volume_ratios.append(ratio)
    log_estimated_volume = sum(np.log(volume_ratios))
    estimated_volume = math.exp(log_estimated_volume) * (math.pi ** (d / 2) / math.gamma(d / 2 + 1))
    return estimated_volume

d = 3
volume_estimate = estimate_volume(d)
print(f"Estimated volume of the {d}-dimensional unit ball: {volume_estimate:.5f}")
```

Metropolis-hasting random walk sampling in 2D

```python
import numpy as np
import matplotlib.pyplot as plt

def is_in_triangle(x, triangle_vertices):
    v0, v1, v2 = triangle_vertices
    def sign(p1, p2, p3):
        return (p1[0] - p3[0]) * (p2[1] - p3[1]) - (p2[0] - p3[0]) * (p1[1] - p3[1])
    d1 = sign(x, v0, v1)
    d2 = sign(x, v1, v2)
    d3 = sign(x, v2, v0)
    has_neg = (d1 < 0) or (d2 < 0) or (d3 < 0)
    has_pos = (d1 > 0) or (d2 > 0) or (d3 > 0)
    return not (has_neg and has_pos)

def metropolis_hastings_walk(triangle_vertices, num_steps, proposal_step=0.1):
    current_point = np.array([0.5, 0.5])
    points = [current_point]

    for _ in range(num_steps):
        proposal = current_point + np.random.uniform(-proposal_step, proposal_step, size=2)
        if is_in_triangle(proposal, triangle_vertices):
            current_point = proposal
        points.append(current_point)
    return np.array(points)

triangle_vertices = np.array([[0, 0], [1, 0], [0.5, 1]])
num_steps = 10000
sampled_points = metropolis_hastings_walk(triangle_vertices, num_steps)

plt.scatter(sampled_points[:, 0], sampled_points[:, 1], s=1, alpha=0.5)
plt.fill(triangle_vertices[:, 0], triangle_vertices[:, 1], 'r', alpha=0.2)
plt.title("Random Walk Inside 2D Triangle")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

```python
import random
import plotly.graph_objects as go

def random_weights_interval_split(n):
    rands = [random.random() for _ in range(n - 1)]
    rands.sort()
    weights = [rands[0]] + [rands[i] - rands[i - 1] for i in range(1, n - 1)] + [1 - rands[-1]]
    return weights

def generate_random_walk(points, weight_func, N):
    generated_points = []
    for _ in range(N):
        weights = weight_func(len(points))
        accum_x = sum(p[0] * w for p, w in zip(points, weights))
        accum_y = sum(p[1] * w for p, w in zip(points, weights))
        accum_z = sum(p[2] * w for p, w in zip(points, weights))
        generated_points.append((accum_x, accum_y, accum_z))
    return generated_points

def plot_tetrahedron_with_walk(points, generated_points):
    Xs, Ys, Zs = zip(*generated_points)
    edge_Xs, edge_Ys, edge_Zs = [], [], []
    for i in range(len(points)):
        for j in range(i + 1, len(points)):
            edge_Xs.extend([points[i][0], points[j][0], None])
            edge_Ys.extend([points[i][1], points[j][1], None])
            edge_Zs.extend([points[i][2], points[j][2], None])

    fig = go.Figure()
    fig.add_trace(go.Scatter3d(x=Xs, y=Ys, z=Zs, mode="markers", marker=dict(size=2, color='blue')))
    fig.add_trace(go.Scatter3d(x=edge_Xs, y=edge_Ys, z=edge_Zs, mode="lines", line=dict(color='black', width=2)))
    fig.show()

points = [(0, 0, 0), (1, 0, 0), (0.5, 0.866, 0), (0.5, 0.2887, 0.816)]
generated_points = generate_random_walk(points, random_weights_interval_split, 10000)
plot_tetrahedron_with_walk(points, generated_points)
```

```python
import numpy as np
import plotly.graph_objects as go
import random

# Define the vertices of the tetrahedron (convex hull)
tetrahedron_vertices = np.array([
    [0, 0, 0],   # Vertex A
    [1, 0, 0],   # Vertex B
    [0, 1, 0],   # Vertex C
    [0, 0, 1],   # Vertex D
])

# 1. Hit-or-Miss Monte Carlo for Sampling Points Inside the Tetrahedron
def hit_or_miss_monte_carlo(tetrahedron_vertices, num_points=10000):
    # Bounding box: Since the tetrahedron is contained in [0, 1]^3, we sample points within this cube
    min_coords = np.min(tetrahedron_vertices, axis=0)
    max_coords = np.max(tetrahedron_vertices, axis=0)

    accepted_points = []

    for _ in range(num_points):
        # Random point in the bounding box
        random_point = np.random.uniform(min_coords, max_coords)

        # Check if the point is inside the tetrahedron
        if is_point_inside_tetrahedron(random_point, tetrahedron_vertices):
            accepted_points.append(random_point)

    return np.array(accepted_points)

def is_point_inside_tetrahedron(point, vertices):
    # Using barycentric coordinates to check if the point is inside the tetrahedron
    v0, v1, v2, v3 = vertices
    v0v1 = v1 - v0
    v0v2 = v2 - v0
    v0v3 = v3 - v0
    point_v0 = point - v0

    # Calculate the dot products
    d00 = np.dot(v0v1, v0v1)
    d01 = np.dot(v0v1, v0v2)
    d02 = np.dot(v0v1, v0v3)
    d11 = np.dot(v0v2, v0v2)
    d12 = np.dot(v0v2, v0v3)
    d22 = np.dot(v0v3, v0v3)
    d03 = np.dot(v0v1, point_v0)
    d13 = np.dot(v0v2, point_v0)
    d23 = np.dot(v0v3, point_v0)

    # Calculate the determinant
    denom = d00 * d11 * d22 + 2 * d01 * d02 * d12 - d00 * d12**2 - d11 * d02**2 - d22 * d01**2

    # Calculate the barycentric coordinates (alpha, beta, gamma, delta)
    alpha = (d11 * d22 * d03 + d01 * d02 * d13 + d02 * d12 * d23 - d01 * d12 * d13 - d02 * d11 * d23 - d03 * d12**2) / denom
    beta = (d00 * d22 * d13 + d01 * d02 * d23 + d02 * d12 * d03 - d01 * d12 * d23 - d02 * d00 * d13 - d03 * d02**2) / denom
    gamma = (d00 * d11 * d23 + d01 * d02 * d03 + d01 * d12 * d13 - d01 * d11 * d13 - d00 * d12 * d23 - d03 * d01**2) / denom
    delta = 1 - alpha - beta - gamma

    # If all barycentric coordinates are between 0 and 1, the point is inside the tetrahedron
    return (alpha >= 0) and (beta >= 0) and (gamma >= 0) and (delta >= 0)

# 2. Markov Chain Monte Carlo (MCMC) to Approximate Uniform Distribution
def mcmc_uniform_distribution(tetrahedron_vertices, num_iterations=10000, initial_point=None):
    # If no initial point, start from a random vertex
    if initial_point is None:
        initial_point = tetrahedron_vertices[np.random.choice(tetrahedron_vertices.shape[0])]

    current_point = initial_point
    points_chain = [current_point]

    for _ in range(num_iterations):
        # Propose a random move: small random step in 3D space
        proposed_point = current_point + np.random.uniform(-0.1, 0.1, 3)

        # Ensure proposed point stays inside the tetrahedron using hit-or-miss method
        if is_point_inside_tetrahedron(proposed_point, tetrahedron_vertices):
            # Accept the proposed point with some probability (Metropolis-Hastings)
            acceptance_prob = 1  # For simplicity, we accept all valid points in this basic version
            if random.random() < acceptance_prob:
                current_point = proposed_point

        points_chain.append(current_point)

    return np.array(points_chain)

# Plotting functions using Plotly for better visualization
def plot_tetrahedron_with_points(tetrahedron_vertices, points, title="Tetrahedron with Points"):
    Xs, Ys, Zs = zip(*points)
    edge_Xs, edge_Ys, edge_Zs = [], [], []
    for i in range(len(tetrahedron_vertices)):
        for j in range(i + 1, len(tetrahedron_vertices)):
            edge_Xs.extend([tetrahedron_vertices[i][0], tetrahedron_vertices[j][0], None])
            edge_Ys.extend([tetrahedron_vertices[i][1], tetrahedron_vertices[j][1], None])
            edge_Zs.extend([tetrahedron_vertices[i][2], tetrahedron_vertices[j][2], None])

    fig = go.Figure()

    # Plot points
    fig.add_trace(go.Scatter3d(x=Xs, y=Ys, z=Zs, mode="markers", marker=dict(size=1, color='blue', opacity=1), name="Points"))

    # Plot edges of the tetrahedron
    fig.add_trace(go.Scatter3d(x=edge_Xs, y=edge_Ys, z=edge_Zs, mode="lines", line=dict(color='black', width=2), name="Tetrahedron Edges"))

    fig.update_layout(title=title,
                      scene=dict(xaxis_title='X', yaxis_title='Y', zaxis_title='Z'),
                      margin=dict(r=10, l=10, b=10, t=40))
    fig.show()

# Run the Hit-or-Miss Monte Carlo Sampling
hit_or_miss_points = hit_or_miss_monte_carlo(tetrahedron_vertices, num_points=10000)
plot_tetrahedron_with_points(tetrahedron_vertices, hit_or_miss_points, title="Hit-or-Miss Monte Carlo Points")

# Run MCMC for Uniform Distribution Approximation
mcmc_points = mcmc_uniform_distribution(tetrahedron_vertices, num_iterations=10000)
plot_tetrahedron_with_points(tetrahedron_vertices, mcmc_points, title="MCMC Points (Uniform Approximation)")
```
