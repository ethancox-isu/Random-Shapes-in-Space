```python
#in this cell we will choose 3 points x1, x2, x3 for which to take the convex 

import plotly.graph_objects as go

N = 10000;

X = [];
Y = [];
Z = [];

#our points are picked to be vertices on the unit cube [0, 1]^3
x1 = [0, 0, 0];
x2 = [1, 0, 0];
x3 = [1, 1, 1];

#ideas 3 to pick the random weights {w_i} to the points {x_i} such that their sum is 1

for i in range(0, N):
    a1 = random.random();
    a2 = random.random();
    a3 = random.random();

    sum = a1 + a2 + a3;

    w1 = a1/sum;
    w2 = a2/sum;
    w3 = a3/sum;

    X = X + [w1*x1[0] + w2*x2[0] + w3*x3[0]];
    Y = Y + [w1*x1[1] + w2*x2[1] + w3*x3[1]];
    Z = Z + [w1*x1[2] + w2*x2[2] + w3*x3[2]];

#plot a 3d scatter plot with plotly

marker = dict(size = 2, opacity = 1);
data = go.Scatter3d(x = X, y = Y, z = Z, mode = 'markers', marker = marker);

margin = dict(l=0, r=0, b=0, t=0);
layout = go.Layout(autosize=True, showlegend=True, margin = margin);

fig = go.Figure(data=data, layout=layout)

fig.update_layout(margin=dict(l=0, r=0, b=0, t=0))
fig.show()
```
